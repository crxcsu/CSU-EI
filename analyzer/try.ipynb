```python
def get_topk(score, ms_index, k: list):
    """
    :param score(N, M): 点积矩阵
    :param ms_index(N, 1): ms索引, 从0开始, 一维, 表示此质谱对应真实smiles在数据库中的索引
    :param k (int): topk
    """
    assert score.shape[0] == ms_index.shape[0]
    # 转换为16位float
    score = score.half()
    score_sorted = torch.argsort(score, dim=1, descending=True)
    tpk = []
    for i in k:
        score_topk = score_sorted[:, :i]
        ms_matched = score_topk == ms_index.view(-1, 1)
        tpk.append(ms_matched.sum()/ms_index.shape[0])
    return tpk
```

```python
# dataset
import numpy as np
import torch
from torch.utils.data import Dataset, DataLoader
import json
from tqdm import tqdm
from model.tokenizer import build_tokenizer
from dataset import *


class GCMSID_100Dataset(Dataset):
    def __init__(self,
                 path,
                 config_json,
                 mode='train'):
        self.mode = mode
        self.data = []
        self.tokenizer = build_tokenizer(config_json['roberta_tokenizer_path'])
        self.config_json = config_json
        self.device = self.config_json['device']

        with open(path, 'r') as f:
            data = json.loads(f.read())
        for tmp_data in data:
            target_data = {'smiles': pre_processing_smi(tmp_data[1], self.tokenizer, config_json),   # [smiles_ids, smiles_mask]  2维
                            'peaks': pre_processing_ms(tmp_data[0], config_json)}  # [ms_tensor, ms_mask]  2维
            self.data.append(target_data)
        print('Finished reading the {} set of NIST Dataset ({} samples found)'.format(mode, len(self.data)))  

    def __getitem__(self, index):
        # Returns one sample at a time
        return self.data[index]["smiles"], self.data[index]["peaks"]  # data是列表，存着字典

    def __len__(self):
        # Return the size of the dataset
        return len(self.data)
```

# 计算于传秀师兄7个质谱的相似度，并绘制热力图

```python
# 读取txt文件里的质谱数据
from typing import List
import torch

def read_ms_txt(ms_data_path):
    with open(ms_data_path, 'r') as f:
        ms_data = f.read()
    peak_data_list = []
    mz_i_list = ms_data.split()
    for i in range(0, len(mz_i_list), 2):
        peak_data_list.append([float(mz_i_list[i]), float(mz_i_list[i+1])])
    return peak_data_list

def process_ms_spectra(ms: List[List[float]], M_max = 1000):
    """
    将输入ms[[m/z, i],[m/z, i],...]转换为tensor[[m/z,..], [intensity,..]],再使用wcs加权得到ms_tensor
    """
    # 1. 将ms转换为tensor[[m/z,..], [intensity,..]]
    # M_max = 1000 最大的m/z值
    ms = torch.tensor(ms, dtype=torch.float32).T # 形状为[2, n] 
    # 2. 将m/z 和 iintensity按照WCS加权
    mz, intensity = ms[0], ms[1]
    intensity = intensity / intensity.max() # 归一化
    ms_ = mz * torch.sqrt(intensity) # WCS加权
    # 3.创建一个维度为M_max的tensor，初始化为0
    ms_tensor = torch.zeros(M_max).float()
    # 用加权后的值填充
    ms_indices = (mz-1).long()
    ms_tensor[ms_indices] = ms_
    
    return ms_tensor

def get_cos_distance(input_1, input_2):
    # 归一化
    input_1 = input_1 / input_1.norm(dim=-1, keepdim=True)
    input_2 = input_2 / input_2.norm(dim=-1, keepdim=True)
    with torch.no_grad():
        return input_1 @ input_2.t()
```

```python
"""
ycx_7_mstest\ms\cl_1062.txt
ycx_7_mstest\ms\cl_1091.txt
ycx_7_mstest\ms\F_905.txt
ycx_7_mstest\ms\F_928.txt
ycx_7_mstest\ms\mo_1738.txt
ycx_7_mstest\ms\mo_1745.txt
ycx_7_mstest\ms\mo_1768.txt
"""
file_path_list = ['ycx_7_mstest/ms/cl_1062.txt', 'ycx_7_mstest/ms/cl_1091.txt', 'ycx_7_mstest/ms/F_905.txt', 'ycx_7_mstest/ms/F_928.txt', 'ycx_7_mstest/ms/mo_1738.txt', 'ycx_7_mstest/ms/mo_1745.txt', 'ycx_7_mstest/ms/mo_1768.txt']
```

```python
wcs_ms_list = []
for file_path in file_path_list:
    ms = read_ms_txt(file_path)
    ms_tensor = process_ms_spectra(ms)
    wcs_ms_list.append(ms_tensor)
# 拼接成一个tensor
wcs_ms_tensor = torch.stack(wcs_ms_list) # [7, 1000]
similarity = get_cos_distance(wcs_ms_tensor, wcs_ms_tensor)
similarity = similarity.cpu().numpy()
```

```python
with open('data/yangqiong_extra_10_test/10compounds_measured_spectra.mgf', 'r') as f:
    data = f.read()

import csv

# 假设你的数据已经以字符串形式存储在变量data中
data = """
BEGIN IONS
COMPOUND_NAME=substance_measuredA.csv
29.0 0.0262 
39.0 0.009399999999999999 
...
END IONS
...
"""

# 分割数据以获取每个分子的数据
molecules_data = data.strip().split("BEGIN IONS")[1:]  # 跳过第一个空元素

# 准备写入CSV
with open('spectra.csv', 'w', newline='') as file:
    writer = csv.writer(file)

    for molecule_index, molecule_data in enumerate(molecules_data):
        # 找到分子名称行和END IONS行
        lines = molecule_data.split('\n')
        compound_name = lines[0].split('=')[1].strip()  # 获取化合物名称
        spectra_data = [line.split() for line in lines[1:-1]]  # 获取光谱数据，去掉首尾行

        # 写入CSV
        writer.writerow([compound_name])  # 写入化合物名称
        for mz, intensity in spectra_data:
            writer.writerow([float(mz), float(intensity)])  # 写入m/z和强度

print("CSV文件已成功写入。")
```

```python
import re
import csv

# 假设文件名为 '10.txt'
file_name = 'data/yangqiong_extra_10_test/10.txt'

# 用于存储所有分子的质谱数据
molecular_data = []
smiles_list = ['CCCCCCOc1ccc(OCCCCCC)cc1',
               'COc1cc(Br)cc(OC)c1',
               'COC(=O)CCCCCBr',
               'COc1c(Br)cccc1C(C)=O',
               'CC(C)(C)c1cc(Br)cc(Br)c1',
               'Brc1ccc(-c2ccc(Br)cn2)nc1',
               'COc1cc(-c2c(OC)cccc2OC)cc(OC)c1OC',
               'CC(C)(C)c1cc(C=O)c(O)c(C=O)c1',
               'c1cc(-c2cc(-c3ccncc3)nc(-c3ccncc3)c2)ccn1',
               'BrCCCCCCCCCCCCOc1cc(Br)cc(Br)c1']

# 读取文件并处理数据
with open(file_name, 'r') as file:
    content = file.read()
    # 按照 "BEGIN IONS" 分割数据
    molecules = content.split("BEGIN IONS")[1:]  # 跳过第一个空元素

    for molecule in molecules:
        # 提取每个分子的m/z和强度
        lines = molecule.strip().splitlines()
        spectra = []
        for line in lines:
            # 使用正则表达式匹配m/z和强度
            match = re.match(r'(\d+\.\d+|\d+)\s+(\d+\.\d+|\d+)', line)
            if match:
                mz = float(match.group(1))  # m/z
                intensity = float(match.group(2))  # 强度
                spectra.append([mz, intensity])
        
        # 将每个分子的光谱数据添加到列表中
        molecular_data.append(spectra)

# 打印结果
for i, data in enumerate(molecular_data):
    print(f"Molecule {i + 1}: {data}")

# 存入csv文件，一列为smiles，一列为质谱数据
with open('data/yangqiong_extra_10_test/10.csv', 'w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(['smiles', 'spectra'])
    for smiles, spectra in zip(smiles_list, molecular_data):
        writer.writerow([smiles, spectra])
```
