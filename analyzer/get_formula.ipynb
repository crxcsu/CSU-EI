```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import rcParams
from collections import Counter
import pandas as pd

rcParams['font.family'] = 'Times New Roman'

df = pd.read_csv('test23/test_unismiles_rank23_class_results.csv')
a = list(df['superclass'])
result = Counter(a)
print (result)
# 去掉类别为nan的
result.pop(np.nan, None)
print (result)
b=dict(result.most_common(10))  # 取前10个
keys = list(b.keys())
print(keys)

# target_df = df[df['subclass'] == target_subclass]

# 4. 定义函数计算 recall@k
def calculate_recall_at_k(data, k):
    total_count = len(data)  # 目标 subclass 的实际总数
    correct_count = len(data[data['rank23'] <= k])  # rank23 小于等于 k 的数量
    if total_count == 0:
        return 0  # 避免除以 0
    return correct_count / total_count
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 数据准备
recall_results = {subclass: [calculate_recall_at_k(df[df['subclass'] == subclass], k) for k in [1, 5, 10]] for subclass in keys}

# 转换为 NumPy 数组
recall_values = np.array(list(recall_results.values()))  # shape: (len(keys), 3)

# 设置雷达图参数
num_classes = len(keys)
angles = np.linspace(0, 2 * np.pi, num_classes, endpoint=False).tolist()
angles += angles[:1]  # 闭合多边形

# 调整图形大小
plt.figure(figsize=(10, 10))  # 更大的图形尺寸
ax = plt.subplot(111, polar=True)

# 绘制每个 k 的 recall 值
recall_labels = ['Recall@1', 'Recall@5', 'Recall@10']
colors = ['#4285f4','#34a853','#db3124']
for i, k in enumerate([0, 1, 2]):  # 对应 recall@1, recall@5, recall@10
    values = recall_values[:, k].tolist()
    values += values[:1]  # 闭合多边形
    ax.plot(angles, values, label=recall_labels[i], color=colors[i], linewidth=2)
    ax.fill(angles, values, color=colors[i], alpha=0.25)

# 设置角度与标签
ax.set_theta_offset(np.pi / 2)
ax.set_theta_direction(-1)

# 设置每个类别的标签
ax.set_xticks(angles[:-1])
ax.set_xticklabels(
    keys, 
    fontsize=16, 
    fontname='Times New Roman', 
    ha='center', 
    va='center', 
    rotation_mode='anchor'
)

# 调整标签和边距
for label, angle in zip(ax.get_xticklabels(), angles[:-1]):
    if angle in [0, np.pi]:  # 顶部和底部的标签
        label.set_horizontalalignment('center')
    elif angle < np.pi:  # 右侧标签
        label.set_horizontalalignment('left')
    else:  # 左侧标签
        label.set_horizontalalignment('right')

# 设置半径标签
ax.set_yticks(np.linspace(0, 1, 6))  # 0 到 1 等分
ax.set_yticklabels([f'{i:.0%}' for i in np.linspace(0, 1, 6)], fontsize=10, fontname='Times New Roman')
ax.set_ylim(0, 1)

# 添加标题与图例
plt.legend(loc='upper right', bbox_to_anchor=(1.5, 1.2), prop={'family': 'Times New Roman','size': 15})
# plt.title('Recall at k Radar Chart', fontsize=14, fontname='Times New Roman')

# 保存与展示
plt.tight_layout(pad=2)  # 增加边距
# plt.savefig('recall_at_k_radar_chart_fixed.png', dpi=300)
plt.show()

```

```python
# 打印不同分子量的分子个数，范围为0~200，200~400，400~600，600~800，800~1000，>1000，打印最大分子量
import pandas as pd
import numpy as np
df = pd.read_csv('test23/test_rank23.csv')
molecular_weight_list = df['molecular_weight'].values
rank23_list = df['rank23'].values
print("max_mw",max(molecular_weight_list))

len_0_100 = len([i for i in molecular_weight_list if 0 < i <= 100])
len_100_150 = len([i for i in molecular_weight_list if 100 < i <= 150])
len_150_200 = len([i for i in molecular_weight_list if 150 < i <= 200])
len_200_300 = len([i for i in molecular_weight_list if 200 < i <= 300])
len_300_400 = len([i for i in molecular_weight_list if 300 < i <= 400])
len_400 = len([i for i in molecular_weight_list if i > 400])

top1_0_100, top1_100_150, top1_150_200, top1_200_300, top1_300_400, top1_400 = 0, 0, 0, 0, 0, 0
top3_0_100, top3_100_150, top3_150_200, top3_200_300, top3_300_400, top3_400 = 0, 0, 0, 0, 0, 0
top5_0_100, top5_100_150, top5_150_200, top5_200_300, top5_300_400, top5_400 = 0, 0, 0, 0, 0, 0
top10_0_100, top10_100_150, top10_150_200, top10_200_300, top10_300_400, top10_400 = 0, 0, 0, 0, 0, 0

for i in range(len(molecular_weight_list)):
    rank = rank23_list[i]
    if 0 < molecular_weight_list[i] <= 100:
        if rank == 1:
            top1_0_100 += 1
        if rank <= 3:
            top3_0_100 += 1
        if rank <= 5:
            top5_0_100 += 1
        if rank <= 10:
            top10_0_100 += 1
    elif 100 < molecular_weight_list[i] <= 150:
        if rank == 1:
            top1_100_150 += 1
        if rank <= 3:
            top3_100_150 += 1
        if rank <= 5:
            top5_100_150 += 1
        if rank <= 10:
            top10_100_150 += 1
    elif 150 < molecular_weight_list[i] <= 200:
        if rank == 1:
            top1_150_200 += 1
        if rank <= 3:
            top3_150_200 += 1
        if rank <= 5:
            top5_150_200 += 1
        if rank <= 10:
            top10_150_200 += 1
    elif 200 < molecular_weight_list[i] <= 300:
        if rank == 1:
            top1_200_300 += 1
        if rank <= 3:
            top3_200_300 += 1
        if rank <= 5:
            top5_200_300 += 1
        if rank <= 10:
            top10_200_300 += 1
    elif 300 < molecular_weight_list[i] <= 400:
        if rank == 1:
            top1_300_400 += 1
        if rank <= 3:
            top3_300_400 += 1
        if rank <= 5:
            top5_300_400 += 1
        if rank <= 10:
            top10_300_400 += 1
    else:
        if rank == 1:
            top1_400 += 1
        if rank <= 3:
            top3_400 += 1
        if rank <= 5:
            top5_400 += 1
        if rank <= 10:
            top10_400 += 1

recall_1 = np.array([top1_0_100, top1_100_150, top1_150_200, top1_200_300, top1_300_400, top1_400]) / np.array([len_0_100, len_100_150, len_150_200, len_200_300, len_300_400, len_400])
recall_3 = np.array([top3_0_100, top3_100_150, top3_150_200, top3_200_300, top3_300_400, top3_400]) / np.array([len_0_100, len_100_150, len_150_200, len_200_300, len_300_400, len_400])
recall_5 = np.array([top5_0_100, top5_100_150, top5_150_200, top5_200_300, top5_300_400, top5_400]) / np.array([len_0_100, len_100_150, len_150_200, len_200_300, len_300_400, len_400])
recall_10 = np.array([top10_0_100, top10_100_150, top10_150_200, top10_200_300, top10_300_400, top10_400]) / np.array([len_0_100, len_100_150, len_150_200, len_200_300, len_300_400, len_400])
print("recall_1", recall_1)
print("recall_3", recall_3)
print("recall_5", recall_5)
print("recall_10", recall_10)
```

# 把nist23所有数据的inchikey存入xlsx

```python
import json
import tqdm

# 把mainlib23中的所有inchikey和SMILES存入一个xlsx文件中，列名为INCHIKSY与SMILES
import pandas as pd
inchikey_list = []
smiles_list = []


for j in tqdm.tqdm(range(70)):
    main_path = 'mainlib23_right_smiles_uni/' + str(j*5000+1) + '-' + str(j*5000+5000) + '.json'
    with open(main_path, 'r') as df:
        main_data = json.load(df)
    for i in main_data:
        inchikey_list.append(i["InChIKey"])
        smiles_list.append(i["smiles"])
    if j == 39:
        # 创建DataFrame
        df = pd.DataFrame({'INCHIKEY': inchikey_list, 'SMILES': smiles_list})
        # 保存为Excel文件
        df.to_excel('mainlib23_inchikey_smiles_1.xlsx', index=False)
        inchikey_list = []
        smiles_list = []
    if j == 69:
        # 创建DataFrame
        df = pd.DataFrame({'INCHIKEY': inchikey_list, 'SMILES': smiles_list})
        # 保存为Excel文件
        df.to_excel('mainlib23_inchikey_smiles_2.xlsx', index=False)
```
